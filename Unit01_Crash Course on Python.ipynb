### 03 -  猜數字（Guess the number）
#### 請使用者猜一個介於範圍（例如 1 到 100）之間的數字，
#### 若使用者猜錯，就詢問他們想繼續玩還是退出；若使用者猜對，就顯示祝賀訊息，
#### 並統計使用者的嘗試次數。如果使用者輸入的數字超出設定範圍，就顯示錯誤提示。

![](data\guess.png)


````python
import random #載入random模組可以隨機生成數字
Max = 100 #程式變數初始化
mini = 0  #程式變數初始化
#ans = 45
ans = random.randint(1,99) #randint 是 random integer的縮寫,此處的終止值是有包含的(跟上面的range不同之處)(想像成輸入2,會從0開始執行0、1、2)
count = 0 #初始指令
while True: #不知道執行次數用while迴圈
    guess = int(input(f"請輸入 {mini} < ? < {Max} 範圍的數字：")) #print這一行也在放在while迴圈內(也要重複執行)
    #f-string語法:f"純文字 {你的變數} 純文字"
    if guess < Max and guess >mini: #設定guess要在mini<?<Max的範圍內條件式
    else:
        print("out of range")
    count += 1 #count = count + 1
    if guess == ans :
        print("Bingo,you are right!")
        break
    elif guess > ans :
        Max = guess
        print("猜太大!你猜了%d次"%(count)) #%d:整數變數,%f:浮點數,%s:字串。補充:此為c語言延伸之
    else : #else後面不能加判斷條件，要淨空，否則會出現syntax error
        mini = guess
        print("猜太小!你猜了%d次"%(count))
````

````python
# Get user email address
email = input("What is your email address?: ").strip()#.strip用來移除整個字串前後的空格

# Slice out the user name,@前的字
user_name = email[0:email.index('@')]
# Slice the domain name,@後的字"
domain_name = email[email.index('@')+1:len(email)]#不能寫成domain_name = email[email.index('@')+1:-1]因 Python 的切片不包含結尾位置，會少算一個。

# Format message
output = "Your username is '{}' and your domain name is '{}'".format(user_name,domain_name)

# Display output message
print(output)


domain = {'gmail.com':'Google','yahoo.com.tw':'Yahoo','ntu.edu.tw':'臺大','hotmail.com.tw':'Hotmail','hotmail.com':'Hotmail'}
if domain_name in domain: #如果domain_name有符合字典中任何一key,就為True然後進入到下一行
    output = "這是註冊在 {} 之下的 Email 地址'".format(domain[domain_name]) #字串格式化用法:{} 和 format的應用,參考A-3,p2。字典召喚咒語:字典[某key]
else:
    
    output = "這是在 {} 之下自定義域".format(domain_name) #else下的domain_name 不在domain字典中
print(output)
````

## 09 - 回文判斷（Is a palindrome）
#### 請使用者輸入單字，判斷它是否為回文，也就是該單字前後對稱，例如 madam，從前讀到後或是從後讀到前的順序都是 madam。

> 題目：<br>
輸入(Input): 雨 滋 春 樹 碧 連 天  天 連 碧 樹 春 滋 雨 <br>
輸出(Output):The text you have entered is a palindrome!<br>
輸入(Input): 資工訓練班 <br>
輸出(Output):The text you have entered is not a palindrome.


````python
def convertInputString():
    #請使用者輸入一串字串，先轉換字串為統一小寫
    rawInput = input("\nPlease enter a word, phrase, or a sentence \nto check if it is a palindrome: ") 
    rawString = rawInput.lower()
    rawList = list(rawString) #有回傳值,要assign變數去接回傳值
    return rawList

def stripAnalphabetics(dirtyList): 
    analphabeticList = [" ", "-", ".", ",", ":", ";", "!", "?", "'", "\""] 
    #去除相關標點符號
    #只有string才能用.replace,這邊是list所以用.remove
    #直到dirtyList內沒有任何標點符號後才會離開迴圈
    for character in analphabeticList:
        if character in dirtyList:
            dirtyList.remove(character)
            return stripAnalphabetics(dirtyList)
        return dirtyList

def runPalindromeCheck(straightList):
    reversedList = straightList[::-1] 
    if reversedList ==  straightList[::-1] :
        return "The text you have entered is a palindrome!" 
    else: 
        return "The text you have entered is not a palindrome." 

def main(): 
    print("\nPalindrome checker") 
    originalList = convertInputString()
    originalList = stripAnalphabetics(originalList)
    palindromeCheck = runPalindromeCheck(originalList)
    print(palindromeCheck)

#def自定義函式,但都部會先執行,直到呼叫方式:函式()
main() 
````

````python
fin_E = open('data/english_list.csv',"r",encoding='UTF-8') #打開檔案用open函式#r:讀檔
fin_M = open('data/math_list.csv',"r",encoding='UTF-8')
lisE=[]
lisM=[]
name=[]

for line in fin_E.readlines(): #readlines:多行讀取
    #print(line,end='')#把多的空行刪除 #csv:欄位間會以","隔開。每換一行就是新的一筆資料。
    line = line.strip().split(",") #strip把多餘空格去除
    #print(line)
    name.append(line[0])
    lisE.append(line[1])
     
for line in fin_M.readlines():
    line = line.strip().split(",")
    lisM.append(line[1])
    
score=[]
fout = open("data/Score.csv","w") #打開Score.csv檔案進行寫入,w:寫
line=''
for i in range(1,len(lisE)):
    score.append(int(lisE[i])+int(lisM[i]))
    list1 = [name[i],str(score[i-1]),"\n"]
    print(name[i],str(score[i-1]))
    
    line = ",".join(list1) #變回字串
    fout.write(line) #寫入
fin_E.close()  #一定要記得關檔(開檔→讀取寫入→關檔),否則要刪除或是對檔案做其他事情都會無法進行
fin_M.close()
fout.close()

'''
line=''
fout = open("Score.csv","r")
for i in (len(name)):
    print()
'''
````
